```python
import requests
import pandas as pd
import time

base_url = "https://bipadportal.gov.np/api/v1/incident/"

all_records = []
offset = 0
limit = 1000

while True:
    url = f"{base_url}?limit={limit}&offset={offset}"

    response = requests.get(url, timeout=60)
    response.raise_for_status()

    data = response.json()
    records = data.get("results", [])

    print(f"Offset {offset}: {len(records)} records")

    if not records:
        break

    all_records.extend(records)
    offset += limit

    time.sleep(0.5)

print("Total records downloaded:", len(all_records))
```

```python
df = pd.DataFrame(all_records)

df.shape
```

```python
import requests
import time

base_url = "https://bipadportal.gov.np/api/v1/loss/"

all_losses = []
offset = 0
limit = 10000

while True:
    url = f"{base_url}?limit={limit}&offset={offset}"
    
    response = requests.get(url, timeout=60)
    response.raise_for_status()
    
    page = response.json()["results"]
    
    if not page:
        break
    
    all_losses.extend(page)
    
    print(f"Downloaded: {len(all_losses)} loss records")
    
    offset += limit
    time.sleep(0.2)

print("Total loss records:", len(all_losses))
```

```python
loss_df = pd.DataFrame(all_losses)

print(loss_df.shape)
loss_df.head()
```

```python
loss_df.columns.tolist()
```

```python
print("Incident rows:", len(df))
print("Loss rows:", len(loss_df))

print("\nIncident loss IDs:")
print(df["loss"].head(10))

print("\nLoss IDs:")
print(loss_df["id"].head(10))
```

```python
print("Unique loss IDs:", loss_df["id"].nunique())
print("Duplicate loss IDs:", loss_df["id"].duplicated().sum())
```

```python
incident_loss_ids = df["loss"].dropna()

matched = incident_loss_ids.isin(loss_df["id"])

print("Incidents with a loss ID:", len(incident_loss_ids))
print("Matching loss IDs:", matched.sum())
print("Unmatched loss IDs:", (~matched).sum())
```

```python
merged_df = df.merge(
    loss_df,
    how="left",
    left_on="loss",
    right_on="id",
    suffixes=("_incident", "_loss"),
    validate="one_to_one"
)
```

```python
print("Original incident rows:", len(df))
print("Merged rows:", len(merged_df))
print("Merged columns:", len(merged_df.columns))
```
